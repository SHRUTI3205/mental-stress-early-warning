# Mental Stress Analytics — Data Exploration

This notebook explores the modeling dataset generated from StudentLife EMA stress, sleep, and activity responses.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path
df = pd.read_csv(Path('..')/'data/processed/model_dataset.csv')
df.head()
```

```python
print(df.shape)
print(df.dtypes)
display(df.isna().mean().sort_values(ascending=False).head(15))
```

```python
display(df['stress_level'].value_counts().sort_index())
sns.countplot(data=df,x='stress_level')
plt.title('Stress EMA response distribution')
plt.show()
```

```python
display(df.groupby('stress_flag')[['sleep_hours','sleep_quality','working','relaxing','social_activity']].mean().round(2))
```
